# Задача 1 – Tracer

## Код

```python
def tracer(func):
    depth = 0
    steps = []

    def traced(arg):
        nonlocal depth
        if depth == 0:
            steps.clear()
        steps.append('  ' * depth + 'вызов ' + str(arg))
        depth += 1
        result = func(arg)
        depth -= 1
        steps.append('  ' * depth + 'возврат ' + str(result))
        return result

    traced.steps = steps
    return traced
```

## Идея

Заводим декоратор tracer, внутри храним глубину depth и список steps. Перед вызовом записываем аргумент с отступом и увеличиваем глубину на 1

После вызова уменьшаем глубину и записываем результат. Таким образом, вызов и возврат одного уровня будут с одинаковым отступом. При новом запуске очищаем steps. Декоратор принимает функцию с одним аргументом

Дополнительная работа без записи шагов: **O(k)** на k вызовов. Если суммарная длина шагов L, время — **O(k + L)**, память — **O(h + L)**, где h — глубина рекурсии. Сложность самой функции считаем отдельно

## Визуализация

![Схема](schema.png)

Считаем factorial(3). При спуске увеличиваем отступ, при возврате уменьшаем. В steps остается последовательность вызовов и результатов

```python
@tracer
def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)

factorial(3)
factorial.steps
```

## Тесты

```python
def test_tracer():
    @tracer
    def factorial(n):
        return 1 if n == 0 else n * factorial(n - 1)

    assert factorial(2) == 2
    assert factorial.steps == [
        'вызов 2', '  вызов 1', '    вызов 0',
        '    возврат 1', '  возврат 1', 'возврат 2']
    assert factorial(0) == 1
    assert factorial.steps == ['вызов 0', 'возврат 1']
    print('OK')


test_tracer()
```
