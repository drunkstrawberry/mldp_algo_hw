# Задача 3 – простые числа

## Код

```python
import pytest
```

```python
def count_prime_nums(n: int) -> int:
    if n<=0:
        return 0
    is_prime_nums = [True] * n
    for i in range(2, n):
        if is_prime_nums[i]:
            for j in range(i ** 2, n, i):
                is_prime_nums[j] = False
    cnt = 0
    for k in is_prime_nums:
        if k:
            cnt += 1
    return cnt - 1
        
```

## Идея алгоритма

Используем решето Эратосфена. Создаем булевый массив is_prime_nums из n элементов, определяющий, является число простым или нет. По умолчанию все числа считаем простыми. Поскольку нам нужно найти все числа СТРОГО меньшие n, нам подходит массив из n элементов, иначе элементов было бы n+1.Пропускаем 0 и 1 и идем циклом от 2 до n. 

Если элемент массива по индексу i True, то число i простое, и тогда с помощью цикла с шагом i мы проставляем в массиве is_prime_nums для всех индексов, производных от i значение False (так как эти числа уже сложные, поскольку делятся на i). Очевидно, что первым числом, производным от i и не проставленным заранее меньшими значениями i будет число i^2. 

Например: у нас есть простое число 5 и тогда его производные 10, 15, 20 будут результатом умножения пятерки на числа меньше, и эти производные уже проставлены, когда в цикл попадали меньшие числа 2 и 3 соответственно. Тогда первым числом, не проставленным как сложное для числа 5 будет число 25, то есть 5^2.

После заполнения массива is_prime_nums необходимо посчитать количество простых чисел. Реализован простой подсчет в цикле и при выходе из количества чисел вычитается 1, так как число 0 не является простым числом.

Также, учитывая ограничения задачи, на вход может быть подано отрицательное целое число или 0. В таком случае программа должна вывести 0, так как отрицательные числа по умолчанию не являются простыми.

Асимптотическая сложность: **O(n log(log(n)))**

## Тесты

```python
def test_count_prime_nums():
    assert count_prime_nums(7) == 4
    assert count_prime_nums(2) == 1
    assert count_prime_nums(1) == 0
    assert count_prime_nums(0) == 0
    assert count_prime_nums(-1) == 0
    print('ОК')
```
