# Задача 1 – число-палиндром

## Код

```python
import pytest
```

```python
def is_num_palindrome(n: int) -> bool:
    a = []
    while n != 0:
        a.append(n % 10)
        n = n//10
    left, right = 0, len(a)-1
    fl = True
    while left <= right:
        if a[left] != a[right]:
            fl = False
        left += 1
        right -= 1
        if not fl:
            return fl
    return fl
```

## Идея алгоритма

Идейно, задача определения, является ли последовательность палиндромом решается с помощью двух указателей. Указатели ставятся на первый и последний элементы последовательности и двигаются к центру последовательности, пока не встретят разные элементы на "зеркальных" индексах, в этом случае выводится False, если таких элементов не нашлось, то выводится True

В данном примере такое решение требовало модификации ввиду ограничения на использование строк и подачи на вход целого положительного числа, которое не является последовательностью. Было выбрано использовать в качестве последовательности массив, в который нужно было последовательно записать цифры. Благодаря тому, что мы решаем задачу определения палиндрома, нам не было критично записать число "задом наперед", последовательно записывая остаток от деления исходного числа на 10 в конец массива и оставляя во входном числе результат целочисленного деления на 10.

После преобразования исходного числа в массив был реализован стандартный алгоритм определения палиндрома.

Асимптотическая сложность: **O(n)**

## Тесты

```python
def test_positives():
    assert is_num_palindrome(4) is True
    assert is_num_palindrome(88) is True
    assert is_num_palindrome(404) is True
    assert is_num_palindrome(2002) is True
    assert is_num_palindrome(12321) is True
    print('OK')

def test_negatives():
    assert is_num_palindrome(30) is False
    assert is_num_palindrome(67) is False
    assert is_num_palindrome(365) is False
    assert is_num_palindrome(2007) is False
    assert is_num_palindrome(50000) is False
    print('OK')
```

```python
test_positives()
test_negatives()
```
