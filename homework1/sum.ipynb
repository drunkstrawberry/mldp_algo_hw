# Задача 2 – сумма

## Код

```python
import pytest
```

```python
def max_even_sum(numbers: list) -> int:
    min_odd = None
    sum = 0
    for i in range(len(numbers)):
        if numbers[i] % 2 != 0 and (min_odd is None or numbers[i] < min_odd):
            min_odd = numbers[i]
        sum += numbers[i]
    if sum % 2 != 0:
        return sum - min_odd
    else:
        return sum
```

## Идея алгоритма

Сумма всех четных чисел делится на 2, а также сумма четного количества нечетных чисел тоже делится на 2. Очевидно, что таким образом в максимальной сумме, которая бы делилась на 2 должны участвовать или все числа в массиве (если нечетных четное количество), или все числа, кроме минимального нечетного. Таким образом, алгоритм реализуется суммированием всех чисел в массиве и проверкой деления суммы на 2, если сумма на 2 не делится, то из общей суммы чисел вычитается минимальное нечетное число

Асимптотическая сложность: **O(n)**

## Тесты

```python
def test_max_even_sum():
    assert max_even_sum([1, 3, 5]) == 8
    assert max_even_sum([2, 4, 6]) == 12
    assert max_even_sum([1]) == 0
    assert max_even_sum([2]) == 2
    assert max_even_sum([1, 9, 2, 8, 3]) == 22
    print('OK')
```
