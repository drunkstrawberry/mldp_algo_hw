# Задача 2 – Validate

## Код

```python
import pytest
```

```python
class Node:
    def __init__(self, val, next):
        self.val = val
        self.next = next
```

```python
class Stack:
    def __init__(self):
        self.head = None

    def push(self, val):
        new_node = Node(val, self.head) 
        self.head = new_node

    def pop(self):
        if self.head is None:
            return None

        val = self.head.val
        self.head = self.head.next
        return val

    def peek(self):
        if self.head is None:
            return None
        return self.head.val

    def to_list(self):
        cur = self.head
        l = []
        while cur is not None:
            l.append(cur.val)
            cur = cur.next
        return l
```

```python
def validate(pushed, popped) -> bool:
    s = Stack()
    j = 0
    for i in range(len(pushed)):
        s.push(pushed[i])
        while j < len(popped) and s.peek() == popped[j]:
            s.pop()
            j += 1
    if s.peek() is None:
        return True
    return False
```

## Идея

Для того чтобы вывести True все элементы pushed должны быть запушены, а все элементы popped должны быть удалены. Поскольку pushed является перестановкой popped, а оба массива содержат уникальные числа, а push и pop в стеке задействует один и тот же "верхний" элемент, на выходе должен получиться пустой стек

Заводим пустой стек и с помощью цикла пушим все элементы pushed. Oперация push в этом кейсе не имеет ограничений, мы можем выполнять ее с любыи элементом. После каждого запушенного элемента pushed мы удаляем все подходящие элементы массива popped, которые можем удалить. Эта операция уже с условием, мы одновременно с помощью индекса проверяем, не закончился ли массив popped, а также проверяем соответствие верхнего элемента стека рассматриваемому элементу popped.

В конце мы знаем, что все элементы pushed запушены и, чтобы проверить, удалены ли они соответственно с popped, мы проверяем, пустой ли остался стек



Асимптотическая сложность алгоритма: **O(n)**

Объем потребляемой памяти: **O(n)**

## Тесты

```python
def test_validate():
    assert validate([1], [1]) is True #push(1), pop(1)
    assert validate([1, 2, 3, 4, 5], [1, 3, 5, 4, 2]) is True
    assert validate([1, 2, 3], [3, 1, 2]) is False
    assert validate([1, 2, 3], [1, 3, 2]) is True
    print('OK')
```
