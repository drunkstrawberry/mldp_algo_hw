# Задача 3 – Merge lists

## Код

```python
class Node:
    def __init__(self, val, next):
        self.val = val
        self.next = next
```

#### С использованием фиктивного элемента

```python
def merge_lists_dummy(list1, list2):
    dummy = Node(None, None)
    tail = dummy

    while list1 is not None and list2 is not None:
        if list1.val <= list2.val:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next
        tail = tail.next

    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return dummy.next
```

#### Без фиктивного элемента

```python
def merge_lists(list1, list2):
    if list1 is None:
        return list2
    if list2 is None:
        return list1

    if list1.val <= list2.val:
        head = list1
        list1 = list1.next
    else:
        head = list2
        list2 = list2.next
    tail = head

    while list1 is not None and list2 is not None:
        if list1.val <= list2.val:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next
        tail = tail.next

    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return head
```

## Идея

Для обоих способов сначала заводим класс узла (Node), в котором определяем значение val и ссылку на следующий узел next. В функции передаем головы двух отсортированных списков list1 и list2, пустой список задаем как None.

Сравниваем первые оставшиеся элементы двух списков и добавляем меньший в конец результата. Так как исходные списки отсортированы, этот элемент будет наименьшим из всех оставшихся. При равных значениях берем элемент из list1.

Для добавления элемента перезаписываем ссылку tail.next на выбранный узел, затем сдвигаем соответствующий list1 или list2 на следующий узел. Сам tail тоже сдвигаем на добавленный элемент.

Когда один из списков закончился, в tail.next записываем оставшуюся часть второго списка. Она уже отсортирована, поэтому перебирать ее по одному элементу не нужно.

Новые узлы для значений не создаем, меняем ссылки между исходными узлами. После слияния исходные списки уже нельзя использовать как две независимые цепочки. Предполагаем, что на входе списки без циклов и общих узлов.

#### С использованием фиктивного элемента

Заводим фиктивный узел dummy, его значение не участвует в сравнении. В tail записываем dummy, пока результат пуст.

Первый выбранный узел записываем в dummy.next так же, как добавляем все остальные элементы через tail.next. Поэтому отдельно выбирать голову списка не нужно.

В конце возвращаем dummy.next, так как сам фиктивный элемент в результат не входит. Если оба списка пустые, dummy.next останется None.

#### Без фиктивного элемента

Сначала проверяем, не пустой ли один из списков. Если пустой, сразу возвращаем голову второго. Если оба пустые, вернется None.

Далее отдельно сравниваем первые элементы и записываем меньший в head. Сдвигаем указатель того списка, из которого взяли элемент, а в tail записываем head.

Оставшиеся элементы добавляем тем же циклом, что и в первом способе. В конце возвращаем head.

Асимптотическая сложность обоих алгоритмов: **O(n + m)**, где n и m — длины исходных списков. Каждый выбранный узел обрабатываем один раз.

Объем дополнительной памяти: **O(1)**. Храним только указатели, а в первом способе еще один фиктивный узел. Узлы результата уже есть в исходных списках.

## Тесты

Для удобства тестирования записываем функции to_linked_list и to_list. Они не задействованы в оценке сложности и памяти алгоритмов слияния. Перед каждой проверкой создаем списки заново, так как слияние меняет ссылки между узлами.

```python
def to_linked_list(values):
    head = None
    for val in reversed(values):
        head = Node(val, head)
    return head


def to_list(head):
    cur = head
    l = []
    while cur is not None:
        l.append(cur.val)
        cur = cur.next
    return l
```

```python
def test_merge_lists():
    # проверяем оба способа на одних и тех же примерах
    for merge in (merge_lists_dummy, merge_lists):
        # оба списка пустые
        assert merge(None, None) is None

        # первый список пустой, возвращаем голову второго
        list2 = to_linked_list([1, 2, 4])
        assert merge(None, list2) is list2
        assert to_list(list2) == [1, 2, 4]

        # второй список пустой, возвращаем голову первого
        list1 = to_linked_list([1, 3, 4])
        assert merge(list1, None) is list1
        assert to_list(list1) == [1, 3, 4]

        # пример из условия
        list1 = to_linked_list([1, 2, 4])
        list2 = to_linked_list([1, 3, 4])
        assert to_list(merge(list1, list2)) == [1, 1, 2, 3, 4, 4]

        # списки из одного элемента, голова берется из второго
        list1 = to_linked_list([5])
        list2 = to_linked_list([2])
        assert to_list(merge(list1, list2)) == [2, 5]

        # все значения одинаковые
        list1 = to_linked_list([3, 3])
        list2 = to_linked_list([3])
        assert to_list(merge(list1, list2)) == [3, 3, 3]

        # первый список заканчивается раньше
        list1 = to_linked_list([1])
        list2 = to_linked_list([2, 3, 4])
        assert to_list(merge(list1, list2)) == [1, 2, 3, 4]

        # второй список заканчивается раньше
        list1 = to_linked_list([2, 3, 4])
        list2 = to_linked_list([1])
        assert to_list(merge(list1, list2)) == [1, 2, 3, 4]

        # отрицательные значения и ноль
        list1 = to_linked_list([-5, 0, 2])
        list2 = to_linked_list([-3, -1, 4])
        assert to_list(merge(list1, list2)) == [-5, -3, -1, 0, 2, 4]

    print('OK')
```

```python
def test_original_nodes():
    for merge in (merge_lists_dummy, merge_lists):
        # сохраняем ссылки на исходные узлы
        a3 = Node(4, None)
        a2 = Node(2, a3)
        a1 = Node(1, a2)
        b3 = Node(4, None)
        b2 = Node(3, b3)
        b1 = Node(1, b2)

        head = merge(a1, b1)

        # проверяем именно узлы, а не только их значения
        assert head is a1
        assert a1.next is b1
        assert b1.next is a2
        assert a2.next is b2
        assert b2.next is a3
        assert a3.next is b3
        assert b3.next is None

    print('OK')
```

```python
test_merge_lists()
test_original_nodes()
```
