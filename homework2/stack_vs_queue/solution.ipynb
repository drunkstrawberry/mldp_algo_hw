# Задача 1 – Stack vs queue

## Код

```python
import pytest
```

```python
class Node:
    def __init__(self, val, next):
        self.val = val
        self.next = next
```

```python
class Stack:
    def __init__(self):
        self.head = None

    def push(self, val):
        new_node = Node(val, self.head) 
        self.head = new_node

    def pop(self):
        if self.head is None:
            return None

        val = self.head.val
        self.head = self.head.next
        return val

    def peek(self):
        if self.head is None:
            return None
        return self.head.val

    def to_list(self):
        cur = self.head
        l = []
        while cur is not None:
            l.append(cur.val)
            cur = cur.next
        return l
```

```python
class Queue:
    def __init__(self):
        self.head = None
        self.tail = None

    def peek(self):
        if self.head is None:
            return None
        return self.head.val

    def pop(self):
        if self.head is None:
            return None
        
        if self.head == self.tail:
            val = self.head.val
            self.head = None
            self.tail = None
            return val
        
        val = self.head.val
        self.head = self.head.next
        return val

    def push(self, val):
        new_node = Node(val, None)
        if self.tail is None:
            self.tail = new_node
            self.head = new_node
            return
        
        self.tail.next = new_node
        self.tail = new_node

    def to_list(self):
        cur = self.head
        l = []
        while cur is not None:
            l.append(cur.val)
            cur = cur.next
        return l
```

## Идея

#### Реализация стека и очереди

Для классов стек и очередь сначала заводим клас узла (Node), в котором определяем атрибуты класса: значение и ссылка на следующий узел

В классе стек заводим атрибут head, который будет указывать на верхний элемент. Для пустого стека ставим head = None. Далее реализуем операции peek, push и pop.

В конце записываем функцию to_list для удобства тестирования (она не задействована в оценке сложности и памяти)




#### Операция push

**Стек**

Заводим новый элемент new_node, который объявляем нодой со значением val, переданным в функцию push() и ссылкой на элемент self.head, чтобы self.head стал вторым элементом

Далее перезаписываем self.head на new_node

**Очередь**

Заводим new_node, ноду со значением val и ссылкой на следующий элемент None (тк в очереди этот элемент станет последним)

Далее для элемента, который был последним в очереди до push (self.tail) определяем текущий new_node как next элемент, а сам self.tail перезаписываем на new_node

Отдельно обрабатываем случай пустой очереди, в таком случае на new_node перезаписываем не только self.tail, но и self.head


#### Операция peek

**Стек**

Возвращаем значение self.head, отдельно проверяем, не пустой ли стек. В пустом стеке возвращаем None

**Очередь**

Возвращаем значение self.head, отдельно проверяем, не пустая ли очередь. В пустой очереди возвращаем None


#### Операция pop

**Стек**

Заводим переменную val и записываем туда значение ноды self.head, затем в self.head перезаписываем следующий элемент (self.head.next), теперь он будет первым в стеке, выводим val


**Очередь**

Заводим переменную val и записываем туда значение ноды self.head, затем в self.head перезаписываем следующий элемент (self.head.next), теперь он будет первым в стеке, выводим val

При пустой очереди выводим None, обрабатываем случай с очередью из 1 элемента, тогда self.head будет равен self.tail. В val так же записываем значение self.head и выводим, но одновременно self.head и self.tail определяем как None, так как очередь закончилась.


Асимптотическая сложность операций peek, pop и push: **O(1)** (они задействуют только один элемент последовательности и никак не зависят от длины связного списка)

Объем потребляемой памяти: **O(n)**

## Тесты

```python
def test_stack():
    
    # заводим пустой стек
    s = Stack()
    # попробуем операции peek, to_list. Они должны давать None, если стек пуст
    assert s.peek() is None
    assert s.to_list() == []
    # попробуем удалить элемент операцией рор, она тоже должна вывести None
    assert s.pop() is None

    # теперь добавим в стек 1 элемент
    s.push(5)
    # проверим, что элемент добавлен
    assert s.peek() == 5
    assert s.to_list() == [5]
    # удалим элемент и проверим корректность
    assert s.pop() == 5
    assert s.peek() is None
    assert s.to_list() == []

    # теперь добавим два элемента
    s.push(7)
    s.push(6)
    # проверим
    assert s.peek() == 6
    assert s.to_list() == [6, 7]
    # удалим элемент
    assert s.pop() == 6
    assert s.to_list() == [7]
    assert s.peek() == 7
    print('OK')
```

```python
def test_queue():
    
    # заводим пустую очередь
    q = Queue()
    # попробуем операции peek, to_list. Они должны давать None, если очередь пуста
    assert q.peek() is None
    assert q.to_list() == []
    # попробуем удалить элемент операцией рор, она тоже должна вывести None
    assert q.pop() is None

    # теперь добавим в очередь 1 элемент
    q.push(5)
    # проверим, что элемент добавлен
    assert q.peek() == 5
    assert q.to_list() == [5]
    # удалим элемент и проверим корректность
    assert q.pop() == 5
    assert q.peek() is None
    assert q.to_list() == []

    # теперь добавим два элемента
    q.push(7)
    q.push(6)
    # проверим
    assert q.peek() == 7
    assert q.to_list() == [7, 6]
    # удалим элемент
    assert q.pop() == 7
    assert q.to_list() == [6]
    assert q.peek() == 6
    print('OK')
```

```python
test_stack()
test_queue()
```
