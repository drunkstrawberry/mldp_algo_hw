# Задача 3 – Balanced Binary Tree

## Код

````python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
````

````python
def is_balanced(root):
    def height(node):
        if node is None:
            return 0
        left_height = height(node.left)
        if left_height == -1:
            return -1
        right_height = height(node.right)
        if right_height == -1:
            return -1
        if abs(left_height - right_height) > 1:
            return -1
        return max(left_height, right_height) + 1

    return height(root) != -1
````

## Идея

Заводим функцию height, которая рекурсивно считает высоту поддерева. Для пустого дерева возвращаем 0, для листа получится 1. Сначала получаем высоты левого и правого поддеревьев, потом возвращаем максимальную из них плюс 1

Если высоты отличаются больше чем на 1, возвращаем -1, то есть поддерево не сбалансировано. Если один из детей уже вернул -1, так же передаем -1 наверх, тк для сбалансированности условие должно выполняться для каждого узла

Таким образом, проверяем баланс одновременно с подсчетом высот и не пересчитываем их заново. В конце возвращаем True, если height(root) не равен -1. Значения узлов не сравниваем, тк проверяем форму дерева, а не свойство BST

Асимптотическая сложность алгоритма: **O(n)**, где n — количество узлов. Каждый узел обрабатываем не более одного раза

Объем дополнительной памяти: **O(h)** на рекурсию, где h — высота дерева. Для сбалансированного дерева O(log n), для цепочки O(n)

## Визуализация

```text
       8                      8
      / \                    / \
     4   10                 4   10
    / \                    /
   2   6                  2
                         /
                        1

      True                  False
```

Слева высоты поддеревьев корня равны 2 и 1, у узла 4 — 1 и 1. Справа у узла 4 высоты равны 2 и 0, поэтому дерево не сбалансировано

Пошагово считаем высоты правого дерева, начиная с height(8):

| Шаг | Узел | left_height | right_height | Возвращаем |
| --- | --- | --- | --- | --- |
| 1 | 1 | 0 | 0 | 1 |
| 2 | 2 | 1 | 0 | 2 |
| 3 | 4 | 2 | 0 | -1, разница высот 2 |
| 4 | 8 | -1 | не считаем | -1 |

До листа 1 спускаемся через 8 → 4 → 2 → 1. После возврата -1 из левого поддерева узел 10 уже не обходим. Итог: is_balanced возвращает False

## Тесты

````python
def test_empty_and_small():
    assert is_balanced(None) is True
    assert is_balanced(Node(1)) is True
    assert is_balanced(Node(1, Node(2))) is True
    assert is_balanced(Node(1, None, Node(2))) is True
    assert is_balanced(Node(1, Node(2, Node(3)))) is False
    assert is_balanced(Node(1, None, Node(2, None, Node(3)))) is False
    print('OK')


def test_balanced_trees():
    assert is_balanced(Node(8, Node(4, Node(2), Node(6)), Node(10))) is True
    root = Node(1, Node(2, Node(4), Node(5)), Node(3, Node(6), Node(7)))
    assert is_balanced(root) is True
    # неполное дерево с разницей высот 1
    root = Node(1, Node(2, None, Node(4)), Node(3, Node(5), None))
    assert is_balanced(root) is True
    print('OK')


def test_unbalanced_trees():
    assert is_balanced(Node(8, Node(4, Node(2, Node(1))), Node(10))) is False
    assert is_balanced(Node(8, Node(4), Node(10, None, Node(12, None, Node(14))))) is False
    # у корня высоты равны, но ниже есть нарушение
    left = Node(2, Node(3, Node(4)))
    right = Node(5, Node(6, Node(8), Node(9)), Node(7, Node(10), Node(11)))
    assert is_balanced(Node(1, left, right)) is False
    print('OK')


def test_values_do_not_matter():
    # одинаковые значения и нарушение BST не влияют на баланс
    assert is_balanced(Node(1, Node(1), Node(1))) is True
    assert is_balanced(Node(0, Node(100), Node(-100))) is True
    assert is_balanced(Node(-5, Node(0, Node(-1)), Node(-10))) is True
    print('OK')


def test_chains():
    left = None
    right = None
    for val in range(100):
        left = Node(val, left, None)
        right = Node(val, None, right)
    assert is_balanced(left) is False
    assert is_balanced(right) is False
    print('OK')
````

````python
test_empty_and_small()
test_balanced_trees()
test_unbalanced_trees()
test_values_do_not_matter()
test_chains()
````
