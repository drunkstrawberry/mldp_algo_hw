# Задача 1 – Tree Traversal

## Код

````python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
````

````python
class BST:
    def __init__(self):
        self.root = None

    def insert(self, val):
        self.root = self._insert(self.root, val)

    def _insert(self, node, val):
        if node is None:
            return Node(val)
        if val < node.val:
            node.left = self._insert(node.left, val)
        elif val > node.val:
            node.right = self._insert(node.right, val)
        return node
````

````python
def preorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        result.append(node.val)
        visit(node.left)
        visit(node.right)

    visit(root)
    return result
````

````python
def inorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        visit(node.left)
        result.append(node.val)
        visit(node.right)

    visit(root)
    return result
````

````python
def postorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        visit(node.left)
        visit(node.right)
        result.append(node.val)

    visit(root)
    return result
````

````python
def reverse_preorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        result.append(node.val)
        visit(node.right)
        visit(node.left)

    visit(root)
    return result
````

````python
def reverse_inorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        visit(node.right)
        result.append(node.val)
        visit(node.left)

    visit(root)
    return result
````

````python
def reverse_postorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        visit(node.right)
        visit(node.left)
        result.append(node.val)

    visit(root)
    return result
````

## Идея

Для дерева заводим класс узла Node, в котором определяем значение val и ссылки на левое и правое поддерево left и right. В классе BST храним корень root, для пустого дерева ставим None

При вставке сравниваем val со значением текущего узла. Если val меньше, рекурсивно идем влево, если больше — вправо. Когда дошли до None, создаем новый узел и возвращаем его, чтобы записать ссылку в родителя. Таким образом, слева остаются только меньшие значения, справа — большие. Повторяющиеся значения не добавляем, тк используем строгое BST

В обходах передаем корень дерева и заводим список result. Функция visit рекурсивно обходит поддеревья и добавляет значение текущего узла. Для каждого обхода меняем порядок этих трех действий:

| Обход | Порядок |
| --- | --- |
| preorder | узел, левое, правое |
| inorder | левое, узел, правое |
| postorder | левое, правое, узел |
| reverse preorder | узел, правое, левое |
| reverse inorder | правое, узел, левое |
| reverse postorder | правое, левое, узел |

Каждый узел попадает в result один раз, тк у него один родитель и каждое поддерево вызывается один раз. Для пустого дерева возвращаем пустой список. Порядок reverse-обходов взят из [вики](https://en.wikipedia.org/wiki/Tree_traversal)

Асимптотическая сложность вставки: **O(h)**, где h — высота дерева. Идем по одному пути от корня. Для сбалансированного дерева это O(log n), в худшем случае O(n), где n — количество узлов. Построение дерева из n значений в худшем случае занимает **O(n²)**

Сложность каждого обхода: **O(n)**, каждый узел обрабатываем один раз

Объем дополнительной памяти вставки и обходов: **O(h)** на рекурсию, в худшем случае O(n). Отдельно дерево и список результата занимают **O(n)** каждый. Здесь считаем, что сравнение значений занимает O(1)

## Визуализация

Добавляем значения 8, 3, 10, 1, 6, 14, 4, 7. Затем вставляем 13:

```text
        8                         8
       / \                       / \
      3   10                    3   10
     / \    \                  / \    \
    1   6    14                1   6    14
       / \                       / \   /
      4   7                     4   7 13
```

| Шаг | Текущий узел | Действие |
| --- | --- | --- |
| 1 | 8 | 13 > 8, идем вправо |
| 2 | 10 | 13 > 10, идем вправо |
| 3 | 14 | 13 < 14, идем влево |
| 4 | None | создаем Node(13), записываем в 14.left |

Далее обходим полученное дерево. В таблице каждый шаг — очередное добавление в result, начинаем с пустого списка:

| Шаг | preorder | inorder | postorder | reverse preorder | reverse inorder | reverse postorder |
| --- | --- | --- | --- | --- | --- | --- |
| 1 | 8 | 1 | 1 | 8 | 14 | 13 |
| 2 | 3 | 3 | 4 | 10 | 13 | 14 |
| 3 | 1 | 4 | 7 | 14 | 10 | 10 |
| 4 | 6 | 6 | 6 | 13 | 8 | 7 |
| 5 | 4 | 7 | 3 | 3 | 7 | 4 |
| 6 | 7 | 8 | 13 | 6 | 6 | 6 |
| 7 | 10 | 10 | 14 | 7 | 4 | 1 |
| 8 | 14 | 13 | 10 | 4 | 3 | 3 |
| 9 | 13 | 14 | 8 | 1 | 1 | 8 |

Результат каждого обхода — значения его столбца сверху вниз. Например, в postorder сначала возвращаемся из узлов 1, 4 и 7, затем добавляем 6 и 3. Корень 8 добавляем последним

## Тесты

````python
traversals = (preorder, inorder, postorder,
              reverse_preorder, reverse_inorder, reverse_postorder)


def make_bst(values):
    tree = BST()
    for val in values:
        tree.insert(val)
    return tree


def test_empty_and_single():
    tree = BST()
    assert tree.root is None
    for traversal in traversals:
        assert traversal(tree.root) == []
    tree.insert(0)
    assert tree.root.val == 0
    assert tree.root.left is None and tree.root.right is None
    for traversal in traversals:
        assert traversal(tree.root) == [0]
    print('OK')


def test_traversals():
    tree = make_bst([8, 3, 10, 1, 6, 14, 4, 7, 13])
    # тот же пример, что в визуализации
    expected = (
        [8, 3, 1, 6, 4, 7, 10, 14, 13],
        [1, 3, 4, 6, 7, 8, 10, 13, 14],
        [1, 4, 7, 6, 3, 13, 14, 10, 8],
        [8, 10, 14, 13, 3, 6, 7, 4, 1],
        [14, 13, 10, 8, 7, 6, 4, 3, 1],
        [13, 14, 10, 7, 4, 6, 1, 3, 8],
    )
    for traversal, answer in zip(traversals, expected):
        assert traversal(tree.root) == answer
        # повторный обход не меняет дерево и не накапливает результат
        assert traversal(tree.root) == answer
    print('OK')


def test_insert():
    tree = make_bst([8, 3, 10, 1, 6, 14, 4, 7])
    tree.insert(13)
    assert tree.root.right.right.left.val == 13
    assert tree.root.left.right.left.val == 4
    assert inorder(tree.root) == [1, 3, 4, 6, 7, 8, 10, 13, 14]

    # повторная вставка не создает новые узлы
    root = tree.root
    for val in [8, 1, 6, 13, 14]:
        tree.insert(val)
    assert tree.root is root
    assert inorder(tree.root) == [1, 3, 4, 6, 7, 8, 10, 13, 14]
    print('OK')


def test_chains():
    # только правые и только левые потомки
    values = list(range(100))
    right = make_bst(values)
    left = make_bst(values[::-1])
    for tree in (right, left):
        assert inorder(tree.root) == values
        assert reverse_inorder(tree.root) == values[::-1]
    for traversal in (preorder, reverse_preorder):
        assert traversal(right.root) == values
        assert traversal(left.root) == values[::-1]
    for traversal in (postorder, reverse_postorder):
        assert traversal(right.root) == values[::-1]
        assert traversal(left.root) == values
    print('OK')


def test_negative_values():
    tree = make_bst([0, -3, 2, -5, -1, 1])
    assert preorder(tree.root) == [0, -3, -5, -1, 2, 1]
    assert inorder(tree.root) == [-5, -3, -1, 0, 1, 2]
    assert postorder(tree.root) == [-5, -1, -3, 1, 2, 0]
    assert reverse_preorder(tree.root) == [0, 2, 1, -3, -1, -5]
    assert reverse_inorder(tree.root) == [2, 1, 0, -1, -3, -5]
    assert reverse_postorder(tree.root) == [1, 2, -1, -5, -3, 0]
    print('OK')
````

````python
test_empty_and_single()
test_traversals()
test_insert()
test_chains()
test_negative_values()
````
