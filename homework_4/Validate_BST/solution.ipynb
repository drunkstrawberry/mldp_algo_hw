# Задача 2 – Validate BST

## Код

````python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
````

````python
def validate_bst(root):
    def check(node, lower, upper):
        if node is None:
            return True
        if lower is not None and node.val <= lower:
            return False
        if upper is not None and node.val >= upper:
            return False
        return (check(node.left, lower, node.val)
                and check(node.right, node.val, upper))

    return check(root, None, None)
````

## Идея

В BST все значения слева должны быть меньше текущего узла, а справа — больше. Проверять только детей недостаточно, тк узел может нарушать ограничение своего предка. Используем строгое BST, одинаковые значения не допускаем

Заводим функцию check, в которую передаем узел и границы lower и upper. Если значение не попадает строго между границами, возвращаем False. Далее рекурсивно проверяем левое поддерево с верхней границей node.val и правое с нижней границей node.val. Вторую границу сохраняем, чтобы учитывать ограничения предков

Для корня обе границы None, то есть ограничений нет. Для пустого поддерева возвращаем True. В конце возвращаем результат проверки обоих поддеревьев через and. Таким образом, True получим только если каждый узел соблюдает все ограничения

Асимптотическая сложность алгоритма: **O(n)**, где n — количество узлов. Каждый узел проверяем не более одного раза

Объем дополнительной памяти: **O(h)** на рекурсию, где h — высота дерева. Для сбалансированного дерева O(log n), для цепочки O(n). Сравнение значений считаем за O(1)

## Визуализация

Два дерева отличаются одним значением:

```text
     10                  10
    /  \                /  \
   5    15              5    15
       /  \                 /  \
      12   20              6    20

     True                False
```

Слева 12 больше 10 и меньше 15, поэтому дерево корректное. Справа 6 меньше своего родителя 15, но находится в правом поддереве 10, где все значения должны быть больше 10

Пошагово проверяем правое дерево:

| Шаг | Узел | lower | upper | Действие |
| --- | --- | --- | --- | --- |
| 1 | 10 | None | None | ограничений нет, идем влево |
| 2 | 5 | None | 10 | 5 < 10, оба ребенка None, возвращаем True |
| 3 | 15 | 10 | None | 15 > 10, идем влево |
| 4 | 6 | 10 | 15 | 6 <= 10, возвращаем False |
| 5 | 15 → 10 | — | — | передаем False наверх, результат False |

Узел 20 уже не проверяем, тк and останавливается после False

## Тесты

````python
def test_empty_and_single():
    assert validate_bst(None) is True
    assert validate_bst(Node(0)) is True
    print('OK')


def test_valid_trees():
    assert validate_bst(Node(10, Node(5), Node(15, Node(12), Node(20)))) is True
    assert validate_bst(Node(0, Node(-5, Node(-8), Node(-1)), Node(4))) is True
    # отсутствие искусственных границ для значений
    assert validate_bst(Node(0, Node(-10**30), Node(10**30))) is True
    print('OK')


def test_invalid_trees():
    assert validate_bst(Node(5, Node(6), Node(8))) is False
    assert validate_bst(Node(5, Node(3), Node(4))) is False
    # дети подходят родителю, но нарушают границу корня
    assert validate_bst(Node(10, Node(5), Node(15, Node(6), Node(20)))) is False
    assert validate_bst(Node(10, Node(5, None, Node(12)), Node(15))) is False
    # нарушение границы более далекого предка
    root = Node(10, Node(5, None, Node(8, None, Node(11))), Node(15))
    assert validate_bst(root) is False
    print('OK')


def test_duplicates():
    assert validate_bst(Node(5, Node(5))) is False
    assert validate_bst(Node(5, None, Node(5))) is False
    assert validate_bst(Node(10, Node(5, None, Node(10)))) is False
    print('OK')


def test_chains():
    left = Node(0)
    right = Node(0)
    for val in range(1, 100):
        left = Node(val, left, None)
        right = Node(-val, None, right)
    assert validate_bst(left) is True
    assert validate_bst(right) is True
    assert validate_bst(Node(1, Node(0, Node(2)))) is False
    print('OK')
````

````python
test_empty_and_single()
test_valid_trees()
test_invalid_trees()
test_duplicates()
test_chains()
````
