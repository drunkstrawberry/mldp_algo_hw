# Задача 2 – Anagrams

## Код

```python
def anagrams(strs):
    groups = {}
    for word in strs:
        key = ''.join(sorted(word))
        if key not in groups:
            groups[key] = []
        groups[key].append(word)
    return list(groups.values())
```

## Идея

Поскольку у анаграмм одинаковый набор букв, после сортировки букв в слове у них получится одинаковая строка. Например, для eat, tea и ate после сортировки получится aet. При этом количество каждой буквы тоже учитывается, поэтому aab и abb попадут в разные группы

Заводим словарь groups, в котором ключом будет отсортированное слово, а значением список слов, которые подходят под этот ключ

Далее идем циклом по словам, для каждого слова сортируем буквы и собираем их в строку key. Если такого ключа еще нет, заводим для него пустой список. После этого добавляем текущее слово в список groups[key]

В конце возвращаем значения словаря в виде списка. Порядок групп может отличаться от примера, тк группы добавляются по мере появления слов. В тестах для удобства сравнения сортируем слова и группы. Повторы слов сохраняем, регистр букв учитываем

Асимптотическая сложность алгоритма: **O(n · m log m)** в среднем, где n — количество слов, m — максимальная длина слова (для m ≥ 2). Для каждого слова сортируем буквы, после чего добавляем его в нужную группу

Если слова пустые или из одной буквы, работа занимает O(n). Оценка предполагает, что словарь работает за среднее время, без большого количества коллизий

Объем потребляемой памяти: **O(n + L)**, где L — общее количество букв во всех словах. Храним ключи и списки со ссылками на исходные слова

## Тесты

```python
def normalize(groups):
    # порядок групп и слов внутри группы не влияет на правильность
    return sorted(sorted(group) for group in groups)


def test_anagrams():
    assert normalize(anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat'])) == normalize(
        [['bat'], ['nat', 'tan'], ['ate', 'eat', 'tea']]
    )
    assert anagrams([]) == []
    assert anagrams(['a']) == [['a']]
    assert anagrams(['']) == [['']]
    assert normalize(anagrams(['', '', 'a'])) == normalize([['', ''], ['a']])

    # повторы слов сохраняются, количество каждой буквы важно
    assert anagrams(['ab', 'ab', 'ba']) == [['ab', 'ab', 'ba']]
    assert normalize(anagrams(['aab', 'aba', 'abb'])) == normalize([['aab', 'aba'], ['abb']])

    # нет анаграмм, все слова в одной группе, русские буквы
    assert normalize(anagrams(['cat', 'dog', 'sun'])) == normalize([['cat'], ['dog'], ['sun']])
    assert anagrams(['abc', 'bca', 'cab']) == [['abc', 'bca', 'cab']]
    assert anagrams(['кот', 'ток', 'кто']) == [['кот', 'ток', 'кто']]

    # регистр учитывается
    assert normalize(anagrams(['Ab', 'bA', 'ab'])) == normalize([['Ab', 'bA'], ['ab']])

    # входной список не меняется
    words = ['tea', 'eat', 'bat']
    anagrams(words)
    assert words == ['tea', 'eat', 'bat']
    print('OK')
```

```python
test_anagrams()
```
