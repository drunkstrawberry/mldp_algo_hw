# Задача 1 – Two sum

## Код

```python
def two_sum(arr, k):
    seen = {}
    for i in range(len(arr)):
        need = k - arr[i]
        if need in seen:
            return [seen[need], i]
        seen[arr[i]] = i
    return None
```

```python
answer = two_sum([1, 3, 4, 10], 7)
print(*answer)
```

## Идея

Заводим словарь seen, в который будем записывать значение элемента массива и его индекс. Далее идем циклом по массиву и для каждого элемента считаем need = k - arr[i], то есть число, которого не хватает до нужной суммы

Если need уже есть в словаре, то пара найдена и мы возвращаем индекс этого числа из seen и текущий индекс i. Поскольку в словаре лежат только предыдущие элементы, первый индекс всегда будет меньше второго

Если нужного числа в словаре нет, записываем в него текущий элемент и его индекс. Проверяем наличие пары ДО записи, тк иначе могли бы использовать один элемент два раза. При этом два одинаковых числа на разных индексах подходят, как в примере с двумя пятерками

Индексы считаем с 0, для вывода через пробел используем print(*answer). По условию пара всегда есть, но если ее нет, возвращаем None

Асимптотическая сложность алгоритма: **O(n)** в среднем, где n — длина массива. Для каждого элемента проверка и запись в словарь в среднем занимают O(1)

Объем потребляемой памяти: **O(n)**

Если много ключей попадает в одну корзину словаря, время в худшем случае может вырасти до O(n²). Здесь считаем, что операции с целыми числами занимают O(1)

## Тесты

```python
def test_two_sum():
    # примеры из условия
    assert two_sum([1, 3, 4, 10], 7) == [1, 2]
    assert two_sum([5, 5, 1, 4], 10) == [0, 1]

    # минимальный массив, отрицательные числа, нули
    assert two_sum([2, 7], 9) == [0, 1]
    assert two_sum([-5, -2, 4, 8], -7) == [0, 1]
    assert two_sum([-3, 2, 8], 5) == [0, 2]
    assert two_sum([0, 0], 0) == [0, 1]
    assert two_sum([0, 4, 7], 4) == [0, 1]

    # один элемент нельзя использовать дважды
    assert two_sum([3, 2, 4], 6) == [1, 2]

    # пара в конце массива
    assert two_sum([1, 2, 4, 8], 12) == [2, 3]

    # исходный массив не меняется
    arr = [10, 1, 3, 4]
    assert two_sum(arr, 7) == [2, 3]
    assert arr == [10, 1, 3, 4]

    # вне гарантии условия: пары нет
    assert two_sum([], 0) is None
    assert two_sum([3], 6) is None
    assert two_sum([1, 2], 10) is None
    print('OK')
```

```python
test_two_sum()
```
