# Задача 3 – K-th

## Код

```python
from random import choice


def kth_largest(nums, k):
    if k < 1 or k > len(nums):
        raise ValueError('k должно быть от 1 до длины массива')

    while True:
        pivot = choice(nums)
        greater, less = [], []
        equal = 0
        for val in nums:
            if val > pivot:
                greater.append(val)
            elif val < pivot:
                less.append(val)
            else:
                equal += 1

        if k <= len(greater):
            nums = greater
        elif k <= len(greater) + equal:
            return pivot
        else:
            k -= len(greater) + equal
            nums = less
```

## Идея

Выбираем случайный pivot. Заводим списки greater и less для больших и меньших значений, равные считаем в equal.

Если больших значений хотя бы k, ищем в greater. Если k попадает в группу равных, возвращаем pivot. Иначе вычитаем из k количество больших и равных, тк все они стоят раньше искомого элемента, и продолжаем в less. Повторы считаем отдельными элементами. Исходный массив не меняем.

Асимптотическая сложность алгоритма: **O(n)** в среднем, тк со случайным pivot оставшаяся часть в среднем уменьшается в несколько раз. В худшем случае **O(n²)**, если на каждом шаге убираем один элемент

Объем дополнительной памяти: **O(n)** на списки для разбиения

## Тесты

```python
def test_kth_largest():
    assert kth_largest([3, 2, 1, 5, 6, 4], 2) == 5
    nums = [3, 2, 3, 1, 2, 4, 5, 5, 6]
    assert kth_largest(nums, 4) == 4
    assert kth_largest(nums, 3) == 5
    assert nums == [3, 2, 3, 1, 2, 4, 5, 5, 6]
    assert kth_largest([-3, 0, -1], 3) == -3
    assert kth_largest([7], 1) == 7
    assert kth_largest([2, 2, 2], 2) == 2
    for k in (0, 4):
        try:
            kth_largest([1, 2, 3], k)
        except ValueError:
            continue
        assert False, 'ожидали ValueError'
```

```python
test_kth_largest()
```
