# Задача 2 – Iterative

## Код

```python
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    return result + left[i:] + right[j:]


def mergesort(nums):
    result = nums.copy()
    width = 1
    while width < len(result):
        for start in range(0, len(result), 2 * width):
            mid = start + width
            end = mid + width
            result[start:end] = merge(result[start:mid], result[mid:end])
        width *= 2
    return result


def quicksort(nums):
    result = nums.copy()
    stack = [(0, len(result) - 1)]
    while stack:
        left, right = stack.pop()
        if left >= right:
            continue
        pivot = result[right]
        pos = left
        for i in range(left, right):
            if result[i] < pivot:
                result[pos], result[i] = result[i], result[pos]
                pos += 1
        result[pos], result[right] = result[right], result[pos]
        stack.append((pos + 1, right))
        stack.append((left, pos - 1))
    return result
```

## Идея

В mergesort сначала сливаем соседние куски длины 1. После прохода каждый получившийся кусок уже отсортирован, поэтому удваиваем width и снова сливаем соседей. Последний кусок может быть короче.

В quicksort заводим стек границ. Берем последний элемент участка за pivot, а в pos держим место для следующего меньшего элемента. После прохода ставим pivot в pos. Слева от него все меньше, справа все больше или равно. В стек кладем два участка без pivot, чтобы отсортировать их так же.

**Mergesort:** время O(n log n), тк проходим по массиву при каждом удвоении width. Память O(n) на копию и слияние.

**Quicksort:** на перемешанных различных значениях время в среднем O(n log n), в худшем случае O(n²), например когда pivot всегда крайний. Память O(n): копия массива и до O(n) пар границ в стеке.

## Тесты

```python
def test_sorts():
    for sort in (mergesort, quicksort):
        for nums in ([], [7], [4, 1, 3, 2, 1], [3, -2, 0, -2, 9, 1, -5],
                     [2, 2, 2], list(range(10, 0, -1))):
            original = nums.copy()
            assert sort(nums) == sorted(original)
            assert nums == original
```

```python
test_sorts()
```
